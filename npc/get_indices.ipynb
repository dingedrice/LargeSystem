# Nup indices

```python
# k is the subunit index, starting from 0 to 7
# continuous determines whether the two ends of the indices are merged or separated between NUPs
def indices_list(df_tmp, k = 0, continuous = True):
    start = []
    end = []
    start_tmp = list(df_tmp['CA_start'])
    end_tmp = list(df_tmp['CA_end'])
    if continuous:
        start.append(start_tmp[0])
        for i in range(1, len(start_tmp)):
            if start_tmp[i] == 1+end_tmp[i - 1]:
                continue
            else:
                start.append(start_tmp[i])
                end.append(end_tmp[i-1])
        end.append(end_tmp[-1])
    else:
        start = start_tmp
        end = end_tmp
    return [f"{i+k*N_subunit_CA} to {j+k*N_subunit_CA}" for i, j in zip(start, end)]

def get_indices_list_vmd(df_tmp, k = 0, continuous = True):
    idx =  indices_list(df_tmp, k, continuous)
    return "index " + " ".join(idx)
```

```python
df_tmp = df_chain_info.loc[[31]]
print(get_indices_list_vmd(df_tmp, 7))
```
