# Init

```python
from modeller import *
from modeller.automodel import *    # Load the AutoModel class
import mdtraj as md
import numpy as np

import sys
sys.path.append("/mnt/e/LargeSystem/npc/")
from init import *
```

```python
def reorganize_pdb(pdb_templ, pdb_input, pdb_output):
    input = md.load(pdb_input)
    templ = md.load(pdb_templ)
    atom_list = list(map(str, input.top.atoms))

    match = np.array([], dtype=int)
    for i in templ.top.atoms:
        try:
            match = np.append(match, atom_list.index(str(i)))
        except:
            pass
    templ.xyz[0] = input.xyz[0][match]
    templ.save_pdb(pdb_output)
```

```python
log.verbose()
env = Environ()

# directories for input atom files
env.io.atom_files_directory = ['.', '../atom_files']
```

```python
name = 'd1'
class MyModel(LoopModel):
    def select_loop_atoms(self):
        return Selection(self.residue_range('18:J', '18:J')) 

a = MyModel(env, inimodel = f'{name}.pdb', sequence = name, loop_assess_methods=assess.DOPE)
# a.starting_model= 1
# a.ending_model  = 1

a.loop.starting_model = 1
a.loop.ending_model   = 3
a.loop.md_level       = refine.fast

a.make()
```

```python
name = 'd1'
reorganize_pdb(f'{name}.pdb', f'{name}.BL00030001.pdb', f'{name}_fixed.pdb')
```

```python
name = 'd7'
# Rotate clockwise around the z-axis by 45 degrees, right multiplied
A = Rotate_Z(theta = -np.pi/4)

template = read_template(f'{name}.BL00020001.pdb')
# R = Rotate_Z(k*np.pi/4)
with open (f'{name}_output_rotated.pdb', 'w') as fout:
    for atom in template:
        if atom == "TER":
            fout.write("TER\n")
        else:
            atom.x, atom.y, atom.z = (np.array([atom.x, atom.y, atom.z]) - C_DILATED_A)@A + C_DILATED_A
            fout.write(str(atom) + '\n')

reorganize_pdb(f'{name}.pdb', f'{name}_output_rotated.pdb', f'{name}_fixed.pdb')
```
